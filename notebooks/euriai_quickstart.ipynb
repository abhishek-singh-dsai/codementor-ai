# Euri AI SDK — minimal experiment

This notebook documents the direct SDK call that inspired CodeMentor AI. Run it with your own Euri AI key saved in a local `.env` file. **Never commit credentials or notebook outputs containing private data.**

```python
from dotenv import load_dotenv
from euriai import EuriaiClient
import os

load_dotenv()
api_key = os.getenv("EURIAI_API_KEY")
if not api_key:
    raise RuntimeError("Set EURIAI_API_KEY in your local .env file before running.")

client = EuriaiClient(api_key=api_key, model="gpt-4.1-nano")
```

```python
response = client.generate_completion(
    prompt="Explain recursion to a Python beginner in three steps.",
    temperature=0.4,
    max_tokens=300,
)
print(response["choices"][0]["message"]["content"])
```
